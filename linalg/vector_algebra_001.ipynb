## Аксиомы векторного пространства

Векторное пространство $V$ над полем $F$ (например, над полем вещественных чисел $\mathbb{R}$) — это набор элементов (векторов), для которых определены две операции: **сложение векторов** и **умножение на скаляры** из $F$. Эти операции удовлетворяют следующим условиям (аксиомам):

1. **Ассоциативность сложения**:  
   $\mathbf{u} + (\mathbf{v} + \mathbf{w}) = (\mathbf{u} + \mathbf{v}) + \mathbf{w} \quad \text{для всех } \mathbf{u}, \mathbf{v}, \mathbf{w} \in V.$

2. **Коммутативность сложения**:  
   $\mathbf{u} + \mathbf{v} = \mathbf{v} + \mathbf{u} \quad \text{для всех } \mathbf{u}, \mathbf{v} \in V.$

3. **Нейтральный элемент сложения**:  
   Существует такой элемент $\mathbf{0} \in V$, что  
   $\mathbf{u} + \mathbf{0} = \mathbf{u} \quad \text{для любого } \mathbf{u} \in V.$

4. **Обратный элемент сложения**:  
   Для каждого $\mathbf{u} \in V$ существует такой элемент $-\mathbf{u} \in V$, что  
   $\mathbf{u} + (-\mathbf{u}) = \mathbf{0}.$

5. **Согласованность умножения на скаляр**:  
   $c(\mathbf{u} + \mathbf{v}) = c\mathbf{u} + c\mathbf{v}, \quad (c + d)\mathbf{u} = c\mathbf{u} + d\mathbf{u}$,
   для всех $c, d \in F$, $\mathbf{u}, \mathbf{v} \in V$.

6. **Ассоциативность умножения на скаляр**:  
   $c(d\mathbf{u}) = (cd)\mathbf{u} \quad \text{для всех } c, d \in F \text{ и } \mathbf{u} \in V.$

## Что такое **поле** в контексте векторной алгебры?

В **векторной алгебре** (или линейной алгебре) поле — это **множество чисел**, над которым определено:

* **сложение**, вычитание
* **умножение**, деление (кроме деления на 0)

и эти операции подчиняются определённым свойствам (ассоциативность, коммутативность и т.д.).

---

### Формально:

**Поле** — это множество $F$, для которого определены операции:

* $+ \colon F \times F \to F$
* $\cdot \colon F \times F \to F$

удовлетворяющие следующим свойствам:

1. **Ассоциативность**:
   $$a + (b + c) = (a + b) + c$$
   $$a \cdot (b \cdot c) = (a \cdot b) \cdot c$$

2. **Коммутативность**:
   $$a + b = b + a$$
   $$a \cdot b = b \cdot a$$

3. **Нейтральные элементы**:
   Существует $0$ и $1$ такие, что:
   $$a + 0 = a,$$
   $$a \cdot 1 = a$$

4. **Обратные элементы**:
   Для любого $a \in F$,

   существует $-a$, такое что
   
   $a + (-a) = 0$;

   и если $a \neq 0$, существует $a^{-1}$, такое что
   
   $a \cdot a^{-1} = 1$

5. **Дистрибутивность**:

   $$a \cdot (b + c) = a \cdot b + a \cdot c$$

---

### 📌 Примеры полей:

| Название                | Обозначение                | Примеры                                                  |
| ----------------------- | -------------------------- | -------------------------------------------------------- |
| Поле вещественных чисел | $\mathbb{R}$               | обычные числа: 3.14, -1, 0.5                             |
| Поле комплексных чисел  | $\mathbb{C}$               | $2 + 3i$, $-i$, $\sqrt{2}$                               |
| Поле рациональных чисел | $\mathbb{Q}$               | $\frac{1}{2}, -\frac{3}{4}$                              |
| Конечные поля           | $\mathbb{F}_p$ или $GF(p)$ | Остатки по модулю $p$, например $\mathbb{F}_2 = \{0,1\}$ |

---

### Почему это важно для векторных пространств?

Векторное пространство определяется **над полем** $F$, потому что:

* ты должен уметь **домножать вектор на любое число из $F$**;
* и при этом сохраняются все аксиомы векторного пространства.

Например, обычное пространство $\mathbb{R}^3$ — это **векторное пространство над полем $\mathbb{R}$**.

## В физике и математическом анализе

Например в физике понятие "поле" имеет иной смысл...

### **Поле (физическое/аналитическое)** — это **функция, которая ставит в соответствие каждой точке пространства некоторое значение**.

Есть два главных типа:

1. **Скалярное поле** — каждой точке ставится в соответствие **одно число**.

   > Пример: **температурное поле** — каждой точке пространства ставится в соответствие температура $T = f(x, y, z)$

2. **Векторное поле** — каждой точке сопоставляется **вектор**.

   > Пример: **электрическое поле** — $\vec{E} = f(x, y, z)$ — вектор силы в каждой точке

При этом поле может быть **Стационарным** и **Нестационарным** в зависимости от того - меняется ли оно во времени!

# Что такое вектор?

image.png

Эти примеры физических величин — это отличный способ наглядно понять различие между скалярами и векторами и как-то наглядно показать, что же такое вектор!

Скалярные величины — это такие, которые полностью описываются одним числом. У них нет направления, только величина. Примеры:

* Температура

* Давление

* Объем

* Масса

* Электричекий заряд

* Энергия

Для скаляра не важно, в какую сторону вы его "направите" — у него просто нет направления.

Векторные величины, напротив, описываются не только числом (модулем), но и направлением. Примеры:

* Скорость

* Ускорение

* Сила

* Напряженность электрического поля

Например, зная только, что тело движется со скоростью 5 м/с — этого недостаточно. Нужно знать в каком направлении — и именно это даёт нам вектор.

### Главное отличие:
Скаляр — это просто число (одномерная информация)

Вектор — это направление + величина (многомерная структура)

Разумеется, это лишь иллюстрация для лучшего понимания! Вектор - это куда более сложный для понимания объект в виду того, что математика, которая изучает эти объекты, весьма абстрактна! Однако она очень хорошо помогает описывать физические процессы, и также данный матаппарат крайне необходим в мире машинного обучения.

# Векторное пространство размерности 2.

Если мы рассмотрим плоскость и зададим на ней произвольную систему координат, то в этой системе мы сможем изобразить бесконечное множество векторов различных длин и направлений. Совокупность всех таких векторов, получаемых из начала координат, образует "двумерное векторное пространство".

````python
#@title plot
import matplotlib.pyplot as plt
import numpy as np


np.random.seed(1)
vectors = [np.random.randint(-5, 6, 2) for _ in range(6)]
vectors = [v for v in vectors if not np.all(v == 0)]


origin = np.array([[0, 0]] * len(vectors))
vectors_np = np.array(vectors)
X, Y = vectors_np[:, 0], vectors_np[:, 1]


fig, ax = plt.subplots(figsize=(6, 6))
ax.quiver(*origin.T, X, Y, angles='xy', scale_units='xy', scale=1, color='blue')


ax.set_xlim(-6, 6)
ax.set_ylim(-6, 6)
ax.set_aspect('equal')
ax.grid(True)
ax.axhline(0, color='black', linewidth=0.5)
ax.axvline(0, color='black', linewidth=0.5)

for i, (x, y) in enumerate(vectors):
    ax.text(x + 0.5 * np.sign(x), y + 0.5 * np.sign(y), f'$\\vec{{a}}_{i+1}$', fontsize=14)

plt.title("Векторы $\\vec{a}_1, \\dots, \\vec{a}_6$", fontsize=16)
plt.show()
````

Все показанные на картинке вектора являются представителями двумерного векторного пространства.
$$\overrightarrow{a_1}, \overrightarrow{a_2}, \overrightarrow{a_3}, ..., \overrightarrow{a_6} \in \mathbb{R}^2$$

````python
#@title plot
import numpy as np
import matplotlib.pyplot as plt
import matplotlib.patches as patches
from ipywidgets import FloatSlider, VBox, Output
from IPython.display import display, clear_output

angle_slider = FloatSlider(value=45, min=0, max=360, step=1, description='Angle (°)')
length_slider = FloatSlider(value=5, min=1, max=8, step=0.1, description='Length')

out = Output()

def update_vector_plot(angle_deg, length):
    x = length * np.cos(np.deg2rad(angle_deg))
    y = length * np.sin(np.deg2rad(angle_deg))

    fig, ax = plt.subplots(figsize=(7, 7))
    ax.set_xlim(-9, 9)
    ax.set_ylim(-9, 9)
    ax.set_aspect('equal')
    ax.grid(True)
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)

    ax.quiver(0, 0, x, y, angles='xy', scale_units='xy', scale=1, color='blue')

    ax.plot([x, x], [0, y], 'k--', linewidth=1)
    ax.plot([0, x], [y, y], 'k--', linewidth=1)

    ax.text(0.05, 0.95, f"A = ({x:.2f}, {y:.2f})", transform=ax.transAxes, fontsize=12, va='top')
    ax.text(0.05, 0.90, f"|A| = {length:.2f}", transform=ax.transAxes, fontsize=12, va='top')
    ax.text(0.05, 0.85, f"α = {angle_deg:.1f}°", transform=ax.transAxes, fontsize=12, va='top')

    arc = patches.Arc((0, 0), 3, 3, angle=0, theta1=0, theta2=angle_deg, color='red', linewidth=1.5)
    ax.add_patch(arc)

    plt.title("Rotating Vector with Coordinates", fontsize=14)
    plt.show()

def on_change(change):
    with out:
        clear_output(wait=True)
        update_vector_plot(angle_slider.value, length_slider.value)

angle_slider.observe(on_change, 'value')
length_slider.observe(on_change, 'value')

with out:
    update_vector_plot(angle_slider.value, length_slider.value)

display(VBox([angle_slider, length_slider, out]))
````

## Как задать вектор?

Чтобы задать вектор на плоскости (в двумерном пространстве), есть **два эквивалентных способа**:

### 1. Через модуль и направление

Вектор можно задать, указав:

* **модуль** (или длину) — насколько «длинный» вектор;
* **направление** — угол $\alpha$, который он образует с осью $X$ (в градусах или радианах).

> Это называется **полярная форма** вектора.

Пример: вектор длины 5 и угла $60^\circ$ направлен вверх и вправо, на 60° от оси $X$.

---

### 2. Через координаты

Вектор можно также задать с помощью его **координат** — то есть через пару чисел $(x, y)$, которые указывают, **где находится его конец**, если его начало помещено в точку $(0, 0)$.

> Это называется **декартова (координатная) форма** вектора.

Пример: вектор $\vec{A} = (3, 4)$ означает, что его начало в точке $(0, 0)$, а конец — в точке $(3, 4)$.

---

## Почему это одно и то же?

Если вы знаете:

* угол $\alpha$ и длину $r$, вы можете найти координаты:

$$
\begin{cases}
x = r \cdot \cos(\alpha) \\
y = r \cdot \sin(\alpha)
\end{cases}
$$

Если вы знаете:

* координаты $(x, y)$, вы можете найти модуль и направление:

  $$
  |\vec{A}| = \sqrt{x^2 + y^2}, \quad \alpha = \arctan\left(\frac{y}{x}\right)
  $$

---

## Вывод:

> Вектор можно задать либо через его **геометрические свойства** (длина и направление), либо через **его координаты**. Эти формы — взаимозаменяемы и эквивалентны.




# Сумма векторов (геометрическая и координатная интерпретации)

В **векторном пространстве** сумма векторов — это одна из **базовых операций**, определённая в аксиомах.

Если заданы два вектора:

* $\vec{a} = (a_1, a_2)$
* $\vec{b} = (b_1, b_2)$

то их сумма $\vec{c} = \vec{a} + \vec{b}$ определяется **покомпонентно**:

### Координатная форма суммы:


$$
\vec{a} + \vec{b} =
\begin{pmatrix}
a_1 \\
a_2
\end{pmatrix}
+
\begin{pmatrix}
b_1 \\
b_2
\end{pmatrix}
=
\begin{pmatrix}
a_1 + b_1 \\
a_2 + b_2
\end{pmatrix}
$$


Эта операция определена для любых двух векторов из пространства $\mathbb{R}^2$ и обобщается на $\mathbb{R}^n$.

---

## Геометрическая интерпретация

* Если вектор $\vec{b}$ приложить к концу вектора $\vec{a}$, то сумма $\vec{a} + \vec{b}$ — это вектор, идущий **из начала** $\vec{a}$ **в конец** $\vec{b}$.
* Это называется **правилом треугольника**.

Альтернативно — можно использовать **параллелограмм**: сумма двух векторов — это **диагональ параллелограмма**, построенного на них.

---

## Почему сумма векторов — это вектор?

Потому что:

* операция сложения определена аксиомами векторного пространства,
* результат снова принадлежит тому же пространству (замкнутость по сложению),
* выполняются свойства:

  * $\vec{a} + \vec{b} = \vec{b} + \vec{a}$ (коммутативность)
  * $\vec{a} + (\vec{b} + \vec{c}) = (\vec{a} + \vec{b}) + \vec{c}$ (ассоциативность)

---

## Связь с аксиомами векторного пространства

Напомним: векторное пространство — это множество, где определены две операции:

1. Сложение векторов: $\vec{a} + \vec{b}$
2. Умножение вектора на скаляр: $\lambda \cdot \vec{a}$

Сложение подчиняется аксиомам:

* **Коммутативность**: $\vec{a} + \vec{b} = \vec{b} + \vec{a}$
* **Ассоциативность**: $\vec{a} + (\vec{b} + \vec{c}) = (\vec{a} + \vec{b}) + \vec{c}$
* **Нейтральный элемент**: $\vec{a} + \vec{0} = \vec{a}$
* **Обратный элемент**: $\vec{a} + (-\vec{a}) = \vec{0}$

---

## Вывод:

> Сложение векторов — это **не просто "геометрическая стрелочка"**, а операция, строго определённая **на уровне структуры векторного пространства**, с чёткими аксиомами, законами и алгебраическим смыслом.


````python
#@title plot
import numpy as np
import matplotlib.pyplot as plt
from ipywidgets import FloatSlider, VBox, Output
from IPython.display import display, clear_output

a_x = FloatSlider(value=2, min=-5, max=5, step=0.1, description='a_x')
a_y = FloatSlider(value=1, min=-5, max=5, step=0.1, description='a_y')
b_x = FloatSlider(value=1, min=-5, max=5, step=0.1, description='b_x')
b_y = FloatSlider(value=3, min=-5, max=5, step=0.1, description='b_y')

out = Output()

def update_plot(ax_val, ay_val, bx_val, by_val):
    a = np.array([ax_val, ay_val])
    b = np.array([bx_val, by_val])
    c = a + b

    fig, ax = plt.subplots(figsize=(8, 8))
    ax.set_xlim(-6, 6)
    ax.set_ylim(-6, 6)
    ax.set_aspect('equal')
    ax.grid(True)
    ax.axhline(0, color='black', linewidth=0.5)
    ax.axvline(0, color='black', linewidth=0.5)

    ax.quiver(0, 0, a[0], a[1], angles='xy', scale_units='xy', scale=1, color='blue')
    ax.quiver(0, 0, b[0], b[1], angles='xy', scale_units='xy', scale=1, color='blue')
    ax.quiver(0, 0, c[0], c[1], angles='xy', scale_units='xy', scale=1, color='red')

    ax.plot([a[0], c[0]], [a[1], c[1]], 'k--')
    ax.plot([b[0], c[0]], [b[1], c[1]], 'k--')

    ax.text(a[0] + 0.2, a[1] + 0.2, r'$\vec{a}$', fontsize=14, color='blue')
    ax.text(b[0] + 0.2, b[1] + 0.2, r'$\vec{b}$', fontsize=14, color='blue')
    ax.text(c[0] + 0.2, c[1] + 0.2, r'$\vec{c} = \vec{a} + \vec{b}$', fontsize=14, color='red')

    ax.text(-5.8, 5.3, rf'$\vec{{a}} = ({a[0]:.1f},\ {a[1]:.1f})$', fontsize=12, color='blue')
    ax.text(-5.8, 4.6, rf'$\vec{{b}} = ({b[0]:.1f},\ {b[1]:.1f})$', fontsize=12, color='blue')
    ax.text(-5.8, 3.9, rf'$\vec{{c}} = ({c[0]:.1f},\ {c[1]:.1f})$', fontsize=12, color='red')

    plt.title("Geometric Vector Sum", fontsize=16)
    plt.show()

def on_slider_change(change):
    with out:
        clear_output(wait=True)
        update_plot(a_x.value, a_y.value, b_x.value, b_y.value)

for s in [a_x, a_y, b_x, b_y]:
    s.observe(on_slider_change, 'value')

with out:
    update_plot(a_x.value, a_y.value, b_x.value, b_y.value)

display(VBox([a_x, a_y, b_x, b_y, out]))
````

## 📚 Направляющие косинусы трёхмерного вектора

Пусть задан вектор $\vec{a} = (x, y, z) \in \mathbb{R}^3$.
Обозначим углы между этим вектором и осями координат:

* $\alpha$ — угол между $\vec{a}$ и осью $X$
* $\beta$ — угол между $\vec{a}$ и осью $Y$
* $\gamma$ — угол между $\vec{a}$ и осью $Z$

---

### ✅ Направляющие косинусы

Направляющие косинусы — это косинусы этих углов:

$$
\cos \alpha = \frac{x}{|\vec{a}|}, \quad
\cos \beta = \frac{y}{|\vec{a}|}, \quad
\cos \gamma = \frac{z}{|\vec{a}|}
$$

где $|\vec{a}| = \sqrt{x^2 + y^2 + z^2}$ — длина вектора.

---

### 📐 Теорема: сумма квадратов направляющих косинусов

Для любого ненулевого вектора в трёхмерном пространстве:

$$
\cos^2 \alpha + \cos^2 \beta + \cos^2 \gamma = 1
$$

📌 Это следует из определения:

$$
\left( \frac{x}{|\vec{a}|} \right)^2 + \left( \frac{y}{|\vec{a}|} \right)^2 + \left( \frac{z}{|\vec{a}|} \right)^2
= \frac{x^2 + y^2 + z^2}{x^2 + y^2 + z^2} = 1
$$

---

### 🧮 Формула направляющих косинусов в виде вектора

Можно записать направляющие косинусы в виде вектора-столбца:

$$
\vec{\ell} =
\begin{pmatrix}
\cos \alpha \\
\cos \beta \\
\cos \gamma
\end{pmatrix}
=
\frac{1}{|\vec{a}|}
\begin{pmatrix}
x \\
y \\
z
\end{pmatrix}
$$

---

### 🧠 Зачем знать два угла?

Если известны два направляющих косинуса (например, $\cos \alpha$ и $\cos \beta$), третий можно найти:

$$
\cos \gamma = \sqrt{1 - \cos^2 \alpha - \cos^2 \beta}
$$

или с минусом, если нужно направление вниз.

---

## 🌐 Обобщение на $n$-мерное пространство

Всё, что мы рассмотрели выше, — это **частный случай**.
Для вектора в $\mathbb{R}^n$:

$$
\vec{v} = (v_1, v_2, \dots, v_n)
$$

Направляющие косинусы задаются аналогично:

$$
\cos \theta_i = \frac{v_i}{|\vec{v}|}, \quad i = 1, 2, \dots, n
$$

И аналогичная теорема сохраняется:

$$
\sum_{i=1}^{n} \cos^2 \theta_i = 1
$$

---

### ⚠️ Почему мы не можем "увидеть" $n$-мерный вектор?

Потому что **человеческое восприятие и физическая реальность ограничены 3 измерениями**.
Векторы в $\mathbb{R}^n$ — это **абстрактные математические объекты**, и их свойства изучаются через:

* координаты,
* алгебру,
* скалярные и векторные произведения,
* норму, угол, расстояние и т.д.

---

### 💬 Вывод:

> Направляющие косинусы — это **косинусы углов между вектором и осями**, которые вместе описывают его **направление в пространстве**.
> Эта концепция универсальна: от $\mathbb{R}^3$ до $\mathbb{R}^n$, и лежит в основе анализа геометрии, направлений и ориентации векторов в любой размерности.





# Умножение вектора на скаляр в $\mathbb{R}^n$

### ✅ Определение

Пусть дан вектор $\vec{v} \in \mathbb{R}^n$:

$$
\vec{v} = (v_1, v_2, \dots, v_n)
$$

и скаляр (число) $\lambda \in \mathbb{R}$.

**Произведение вектора на скаляр** — это вектор $\lambda \vec{v} \in \mathbb{R}^n$, компоненты которого равны:

$$
\lambda \vec{v} = (\lambda v_1, \lambda v_2, \dots, \lambda v_n)
$$

---

## 🧮 Формула в виде вектора-столбца

Если $\vec{v}$ задан в виде столбца:

$$
\vec{v} =
\begin{pmatrix}
v_1 \\
v_2 \\
\vdots \\
v_n
\end{pmatrix}
\quad \Rightarrow \quad
\lambda \vec{v} =
\begin{pmatrix}
\lambda v_1 \\
\lambda v_2 \\
\vdots \\
\lambda v_n
\end{pmatrix}
$$

---

### 📐 Геометрический смысл

* Умножение вектора на положительный скаляр **растягивает** его или **сжимает**, сохраняя направление.
* Если $\lambda > 1$ — вектор **удлиняется**
* Если $0 < \lambda < 1$ — **укорачивается**
* Если $\lambda = 0$ — результат: **нулевой вектор**
* Если $\lambda < 0$ — вектор **меняет направление на противоположное**

---

### 📌 Пример

Пусть $\vec{v} = (2, -3, 1)$, тогда:

* $2\vec{v} = (4, -6, 2)$
* $-1\vec{v} = (-2, 3, -1)$
* $0\vec{v} = (0, 0, 0)$

---

## 🧠 Свойства операции $\lambda \vec{v}$

Для любых $\lambda, \mu \in \mathbb{R}$, $\vec{v}, \vec{w} \in \mathbb{R}^n$:

1. **Ассоциативность**:

   $$
   (\lambda \mu) \vec{v} = \lambda (\mu \vec{v})
   $$

2. **Дистрибутивность относительно сложения векторов**:

   $$
   \lambda (\vec{v} + \vec{w}) = \lambda \vec{v} + \lambda \vec{w}
   $$

3. **Дистрибутивность относительно сложения скаляров**:

   $$
   (\lambda + \mu) \vec{v} = \lambda \vec{v} + \mu \vec{v}
   $$

4. **Умножение на 1 и 0**:

   $$
   1 \cdot \vec{v} = \vec{v}, \quad 0 \cdot \vec{v} = \vec{0}
   $$

---

## 🌐 Обобщение

Операция $\lambda \vec{v}$ определена для любого векторного пространства над полем $\mathbb{R}$ или $\mathbb{C}$, и остаётся корректной в любом $n$-мерном случае:

> $$
> \lambda \colon \mathbb{R} \times \mathbb{R}^n \rightarrow \mathbb{R}^n
> $$

---

## ✅ Вывод

> Умножение вектора на скаляр — это фундаментальная операция, которая сохраняет направление (или обращает его), и масштабирует длину вектора.
> Это ключ к линейным преобразованиям, нормализации и формированию линейных комбинаций.

# Скалярное произведение векторов

Скалярное произведение (dot product, inner product) — это операция, при которой двум векторам сопоставляется **одно число** (скаляр), отражающее их **направленную "сходность"** и **взаимную ориентацию**.

---

### 🔷 1. Координатное определение

Пусть заданы два вектора в $\mathbb{R}^n$:

$$
\vec{a} = (a_1, a_2, \dots, a_n), \quad \vec{b} = (b_1, b_2, \dots, b_n)
$$

Их **скалярное произведение** обозначается:

$$
\langle \vec{a}, \vec{b} \rangle = a_1 b_1 + a_2 b_2 + \dots + a_n b_n = \sum_{i=1}^n a_i b_i
$$

---

### 📐 2. Геометрическое определение

Если между векторами $\vec{a}$ и $\vec{b}$ существует угол $\theta$, то скалярное произведение можно выразить через модуль (длину) и угол:

$$
\langle \vec{a}, \vec{b} \rangle = |\vec{a}| \cdot |\vec{b}| \cdot \cos(\theta)
$$

где:

* $|\vec{a}| = \sqrt{a_1^2 + \dots + a_n^2}$ — длина вектора
* $\theta \in [0^\circ, 180^\circ]$ — угол между векторами

---

### 🧮 3. Угол между векторами

Из геометрической формулы можно найти угол между векторами:

$$
\cos(\theta) = \frac{\langle \vec{a}, \vec{b} \rangle}{|\vec{a}| \cdot |\vec{b}|}
\quad \Rightarrow \quad
\theta = \arccos\left( \frac{\langle \vec{a}, \vec{b} \rangle}{|\vec{a}| \cdot |\vec{b}|} \right)
$$

Это удобно использовать для определения **направленной близости** между векторами.

---

## 🧠 Геометрический смысл

* Если $\theta = 0^\circ$ → векторы сонаправлены → $\langle \vec{a}, \vec{b} \rangle > 0$
* Если $\theta = 90^\circ$ → векторы перпендикулярны → $\langle \vec{a}, \vec{b} \rangle = 0$
* Если $\theta = 180^\circ$ → векторы противоположны → $\langle \vec{a}, \vec{b} \rangle < 0$

---

## 🤖 Применение в машинном обучении

### 📏 Косинусная мера (Cosine Similarity)

Скалярное произведение лежит в основе **косинусной близости** — популярной метрики в **NLP, рекомендательных системах, кластеризации**.

$$
\text{cosine_similarity} = \frac{\langle \vec{a}, \vec{b} \rangle}{|\vec{a}| \cdot |\vec{b}|}
$$

📌 Это значение лежит в диапазоне от $-1$ до $1$. Чем ближе к 1, тем **похожее направление** у векторов (например, у текстов, представленных векторно).

---

## ⚙️ Применение в физике

### 💥 Работа силы

Векторная формула для работы:

$$
A = \langle \vec{F}, \vec{d} \rangle = |\vec{F}| \cdot |\vec{s}| \cdot \cos(\theta)
$$

где:

* $\vec{F}$ — сила
* $\vec{s}$ — перемещение

📌 Если сила направлена строго по движению — работа максимальна. Если под углом — уменьшается. Если перпендикулярна — $A = 0$.

---

## ✅ Свойства скалярного произведения

* $\langle \vec{a}, \vec{b} \rangle = \langle \vec{b}, \vec{a} \rangle$ — симметричность
* $\langle \vec{a}, \vec{a} \rangle = |\vec{a}|^2$
* $\langle \lambda \vec{a}, \vec{b} \rangle = \lambda \langle \vec{a}, \vec{b} \rangle$
* Линейность: $\langle \vec{a} + \vec{b}, \vec{c} \rangle = \langle \vec{a}, \vec{c} \rangle + \langle \vec{b}, \vec{c} \rangle$

---

## 🧾 Вывод

> Скалярное произведение — фундаментальная операция, которая объединяет **геометрию** и **алгебру**, лежит в основе **анализа данных, распознавания образов, физики, геометрии и обучения моделей**.


## 🧮 Скалярное произведение в NumPy

Пусть у нас есть два вектора $\vec{a}$ и $\vec{b}$:

```python
import numpy as np

a = np.array([a1, a2, ..., an])
b = np.array([b1, b2, ..., bn])
```

---

### ✅ Способ 1: `np.dot(a, b)`

```python
result = np.dot(a, b)
```

Это возвращает **скаляр** — результат скалярного произведения.

📌 Пример:

```python
a = np.array([1, 2, 3])
b = np.array([4, 5, 6])

result = np.dot(a, b)
print(result)  # 1*4 + 2*5 + 3*6 = 32
```

---

### ✅ Способ 2: `a @ b` (синтаксис Python 3.5+)

```python
result = a @ b
```

Это тоже даёт скаляр (аналог `np.dot` для одномерных массивов).

---

### ✅ Способ 3: `np.inner(a, b)`

```python
result = np.inner(a, b)
```

Для одномерных массивов `np.inner` = `np.dot`.

---

## ⚠️ Важно:

* Все векторы должны быть **одинаковой размерности**.
* Если векторы — **строчные или столбцовые** матрицы (2D), то `np.dot` может дать матрицу, а не скаляр. Поэтому **используй 1D массивы**, если хочешь чистый результат:

```python
a = np.array([1, 2])     # ОК
a = np.array([[1], [2]]) # Не ОК для dot в лоб
```
